### Reconciliation memo — Customer & Marketing domains

**Highest-severity gaps (GDPR-critical):**
- **`customer.email` is an untagged PII column.** Email is personal data, but it carries no
  classification — so it is invisible to any PII-driven access or subject-access workflow.
  This is the first thing the GDPR audit would flag.
- **`campaign_membership.cust_ref` is also untagged PII** — it stores the customer's email as
  the marketing join key. A second ungoverned copy of personal data, in a different domain.

**Table-level gaps:**
- `campaign` has **no owner** — no accountable steward for campaign definitions.
- `campaign_membership` has **no owner and no retention** — ungoverned PII with no deletion
  clock, the worst combination under GDPR.

**Cross-domain semantic conflicts affecting these domains:**
- **Customer identity key:** loyalty keys customers on `customer_id` (integer) while marketing
  keys on `cust_ref` (email). There is no shared key, so customer data can only be reconciled
  across domains by matching on email — fragile and a resolution risk.
- **Marketing consent:** `customer.marketing_consent` (loyalty) and
  `campaign_membership.mkt_consent` (marketing) are two systems of record that disagree. Acting
  on the marketing copy when loyalty has withdrawn consent is a direct GDPR violation.

**Recommended standards:** require `owner`, `classification`, and `retention` on every table
before publish; apply the platform PII tag to `customer.email` and `campaign_membership.cust_ref`
immediately; and designate the loyalty domain as the single consent system of record that
marketing must read rather than copy.
