# Demo — Metadata as a Governance Control

**Trailhead Provisions** runs as a federation of four domain teams on a shared catalog. This
short walkthrough demonstrates the **technique** you'll use in the exercise: treat the catalog
as a *control surface*, run the audit, and learn to read what it's telling you. We'll interpret
a couple of representative findings here — the exercise asks you to do the full audit and write
the reconciliation memo yourself.

```python
import pandas as pd
pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
from governance_toolkit import GovernedCatalog

gc = GovernedCatalog("trailhead.db")
print("Catalog database:", gc.catalog.database)
gc.query("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name")
```

## Run the catalog audit
One row per governance gap the catalog can detect — across all four domains.

```python
audit = gc.catalog_audit()
print('Total gaps:', len(audit))
audit
```

## How to read the audit (two representative findings)

The audit reports two kinds of gap. Here's how to interpret each — you'll find the rest in the
exercise.

**A table-level gap — `orders` has no owner.** Filter to one table to see its gaps:

```python
audit[audit["table"] == "orders"]
```

A missing `owner` means no one is accountable for purchase events — there's no
steward to escalate a quality or policy question to. Missing `classification` or `retention`
would likewise mean undeclared sensitivity or no deletion clock.

**A column-level gap — an untagged PII column.** These are the highest-severity findings,
because a personal-data column with no classification is invisible to PII-driven controls:

```python
audit[audit["gap_type"] == "untagged_column"]
```

Reading one of these: an untagged `email` or `cust_ref` column holds personal data that no PII policy can see — exactly what a GDPR audit flags first.

## Gaps aren't only missing tags — domains also disagree
The audit's companion is the semantic-conflict report: places where domains define the same concept differently.

```python
gc.semantic_conflicts()
```

Read the **consent** conflict: loyalty and marketing each keep their own
`marketing_consent` flag, and they disagree — acting on the wrong copy is a GDPR violation.
That's a governance defect surfaced entirely through metadata, with no row of customer data
required.

**Takeaway / your turn:** the catalog is a control surface — auditing it shows where ownership,
classification, retention, and shared definitions are missing. In the exercise, run this same
audit and write a reconciliation memo for the Customer and Marketing domains.
