# Demo — Auditing a Regulated Number to Source

A CSRD disclosure is only as trustworthy as its traceability to source. This walkthrough runs
the reporting pipeline and reconciles the disclosed figure against a recomputed one; the exercise
asks you to trace a quarter to source and find where integrity breaks.

```python
import pandas as pd
pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
from governance_toolkit import GovernedCatalog

gc = GovernedCatalog("trailhead.db")
rep = gc.run_sustainability_report()
display(rep.quarterly)
rep.reconciliation
```

**Read it:** the disclosed figure (summing `shipment.carbon_kg` blindly) is far
larger than the recomputed, unit-normalized figure — a slice of shipments is stored in grams, not
kilograms, inflating the disclosure every quarter.
