# Demo — Reading Quality Results & Choosing a Strategy

Quality results aren't a pass/fail score — each failure needs a *decision*: fix at the
**source**, fix **downstream** in the transform, or **quarantine** the bad records. This
walkthrough reads one failure and picks a strategy; the exercise asks you to triage them all.

```python
import pandas as pd
pd.set_option("display.width", 160); pd.set_option("display.max_columns", 30)
from governance_toolkit import GovernedCatalog

gc = GovernedCatalog("trailhead.db")
q = gc.run_quality_suite()
print("Failing expectations:", int((~q["success"]).sum()), "of", len(q))
q
```

Roll up by dimension to see where the platform is weakest:

```python
q.groupby("dimension").agg(checks=("success", "size"),
                           failing=("success", lambda s: int((~s).sum())),
                           worst_pct=("unexpected_percent", "max")).reset_index()
```

**Read one failure:** `customer.email` fails *validity* — malformed addresses on a
PII column. Strategy: **quarantine** the bad records (don't send to them, don't return garbage to
a subject-access request) while the producer fixes contact capture.

**Takeaway / your turn:** in the exercise, classify every failing expectation as source /
downstream / quarantine and write a prioritized remediation plan.
