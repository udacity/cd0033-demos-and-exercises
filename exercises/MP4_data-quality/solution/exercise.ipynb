### Prioritized remediation plan

**Priority 1 — Quarantine immediately (`quarantine` — active GDPR & CSRD regulatory harm):**
- **`campaign_membership.mkt_consent` (consistency, `20.71%` / ~21% mismatch across 758 valid-email rows, plus `42` unresolvable `'n/a'` rows in `unresolvable_count`):** Marketing's consent flag disagrees with the Loyalty system of record (`customer.marketing_consent`) on ~21% of resolvable memberships. Contacting customers who withdrew consent is an active GDPR Art. 7/21 violation. **Action:** immediately quarantine all mismatched and unresolvable rows out of `marketing_audience` and campaign sends while re-pointing Marketing to read consent directly from Loyalty.
  - **Target Contract SLO:** `>= 99.0% match loyalty SoR` (`consistency`, owner: `marketing@trailhead.example`).
- **`shipment.carbon_kg` (accuracy, `9.29%` / 65 of 700 rows `> 100 kg`):** One carrier emits shipment weight/emissions in grams (`~1,800–9,500`) instead of kilograms, inflating Trailhead's CSRD quarterly carbon disclosure nearly 100x. **Action:** immediately quarantine rows with `carbon_kg > 100` from the CSRD reporting pipeline while normalizing gram-scaled records (`carbon_kg / 1000.0`) and enforcing unit validation at carrier ingest.
  - **Target Contract SLO:** `100% in kg (0 < kg <= 100)` (`accuracy`, owner: `inventory-fulfillment@trailhead.example`).
- **`customer.email` (`4.55%` / 20 of 440 invalid) and `customer.phone` (`7.95%` / 35 of 440 invalid E.164):** Malformed contact data on PII columns causes failed communications and breaks GDPR subject-access/erasure lookups (including the 42 `'n/a'` join failures in `campaign_membership`). **Action:** quarantine malformed contact records from downstream consumption until corrected at customer signup.
  - **Target Contract SLOs:** `>= 99.5% valid format` on `customer.email` and `>= 99.5% E.164 format` on `customer.phone` (`validity`, owner: `customer-identity@trailhead.example`).

**Priority 2 — Fix at source (`source` — producer contract enforcement):**
- **`product.category` (`8.33%` null) and `product.sustainability_class` (`50.00%` null):** Mandatory product and CSRD classification attributes are missing at catalog ingestion in the Inventory & Fulfillment domain. **Action:** block unpopulated product rows at the producer publish gate and backfill from the ERP master.
  - **Target Contract SLOs:** `>= 99.0% non-null` on both `product.category` and `product.sustainability_class` (`completeness`, owner: `inventory-fulfillment@trailhead.example`).
- **`orders.amount` (`1.11%` / 10 of 900 negative):** Unflagged negative order totals represent an upstream checkout/refund sign bug in the E-commerce Orders pipeline. **Action:** enforce non-negative validation in the Orders producer before emitting events.
  - **Target Contract SLO:** `>= 99.9% non-negative` (`accuracy`, owner: `orders-team@trailhead.example`).

**Priority 3 — Handle downstream (`downstream` — transform & retention rules):**
- **`customer.created_ts` (`5.68%` / 25 of 440 older than 1,095 days):** Stale customer timestamps reflect legacy accounts beyond the 3-year active window (`P3Y`). **Action:** filter or archive accounts older than 1,095 days with no recent order activity in downstream analytical transforms.
  - **Target Contract SLO:** `>= 98.0% within 1095d` (`timeliness`, owner: `customer-identity@trailhead.example`).

**Recurring drift watch (ongoing SLO monitoring):**
- Both the **cross-domain consent mismatch** (`campaign_membership.mkt_consent`, ~21% + 42 unresolvable rows) and the **carrier carbon unit mismatch** (`shipment.carbon_kg`, `9.29%` in grams) are recurring integration drift rather than one-off data corruption. Both must be wired into the governance operations dashboard as continuous, automated SLO gates on every pipeline run.

## Solution Summary & Common Mistakes to Avoid

### Key points
- **Priority 1 (`quarantine` — active GDPR & CSRD harm):** `campaign_membership.mkt_consent` (`20.71%` / ~21% mismatch + 42 unresolvable `'n/a'` rows -> target SLO `>= 99.0% match loyalty SoR`), `shipment.carbon_kg` (`9.29%` gram-scaled rows -> target SLO `100% in kg (0 < kg <= 100)`), `customer.email` (`4.55%` -> target SLO `>= 99.5% valid format`), and `customer.phone` (`7.95%` -> target SLO `>= 99.5% E.164 format`).
- **Priority 2 (`source` — producer contract enforcement):** `product.category` (`8.33%` null -> target SLO `>= 99.0% non-null`), `product.sustainability_class` (`50.00%` null -> target SLO `>= 99.0% non-null`), and `orders.amount` (`1.11%` negative -> target SLO `>= 99.9% non-negative`).
- **Priority 3 (`downstream` — transform/retention):** `customer.created_ts` (`5.68%` > 1095 days -> target SLO `>= 98.0% within 1095d`). Note that `customer.customer_id` passes its `100% unique` (`uniqueness`) SLO.

### Common Mistakes to Avoid
- **Failing to quarantine `campaign_membership.mkt_consent` and `shipment.carbon_kg` in Priority 1** — allowing unconsented marketing rows or 1000x-inflated gram values through to production while waiting on a source refactor creates immediate GDPR and CSRD compliance violations.
- **Quoting a Cartesian-inflated consent mismatch rate** instead of `~21%` (`20.71%` on the 758 valid-email memberships, plus 42 unresolvable `'n/a'` memberships in `unresolvable_count`).
- **Choosing `downstream` for upstream producer bugs** (`orders.amount` negatives and missing `product` attributes belong at `source`).
- **Omitting explicit target contract SLO numbers** from `gc.contract_slos()` for each priority tier.
