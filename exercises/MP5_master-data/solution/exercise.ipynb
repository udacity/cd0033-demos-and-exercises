**Survivorship rule (`survivor="lowest_id"`):** The **lowest `customer_id` wins** as the golden
record — **Customer Identity & Loyalty** is the designated customer system of record and lower IDs
are the original registrations; non-key attributes take the most-recent non-null value.

**False-positive vs false-negative tradeoff:** We accept only pairs where the name and email-base
match is corroborated by phone, country, and registration timestamp (`rows 0..5`), and we **reject**
homonym pairs whose phone, country, and `created_ts` disagree (`rows 6..7`: `Zane Saleh` in `IN` vs
`US`, `Pia Lindqvist` in `CA` vs `BR`). Under GDPR, a **false positive** (over-merging two different
people into one golden record) is far worse than a **false negative** (leaving a duplicate unmerged):
over-merging leaks one customer's PII, order history, and consent state to another person (a
cross-customer data breach under GDPR Art. 5(1)(f) and Art. 32), whereas a missed duplicate only
leaves two records fragmented until more signal arrives.

**Structural cross-domain fix:** Push the loyalty `customer_id` into Marketing (`campaign_membership`)
as the mandatory shared foreign key at capture time, eliminating email-based resolution entirely.
